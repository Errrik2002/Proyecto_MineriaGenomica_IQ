# Instalar Linux/Ubuntu

Ubuntu es una distribucion de Linux, existen varias distribuciones de Linux (ArchLinux, Linuxmint, Ubuntu, Debian y muchas mas), cada distribucion es diferente y unas mas especializadas en aspectos informaticos. Ubuntu es el mas amigable y el mas usado en cuestiones de bioinformatica.

El sistema operativo de Mac, es muy similar a la de Ubuntu, puede usarse esa sin tener que instalar Ubuntu, pero hay comandos diferentes.

## Instalar desde windows

Abrir desde windows, buscar "Windows PowerShell" y correr como administrador (Igual se puede con el Win+R y seleccionar cmd)

```Windows Power shell:
 wsl --install -d ubuntu
```
Con esto se insatalara Ubuntu en tu dispositivo windows. Posterior busca Ubuntu en tu windows y abrelo. Se abrira una terminal y tendras que crear un usuario.

# Instalar Conda en Linux/Ubuntu

Conda es un gestionador de paquetes, para instalar Conda, hay dos opciones: Miniconda y Anaconda.

Anaconda: Se supone que es mas amigable, tiene un visualizador tipo PlayStore, para que puedas ver que puedes descargar, que paquetes, la informacion de esos paquetes. Sin embargo, por lo mismo de la informacion, es mas pesado y lento. 

Miniconda: Miniconda es la version minima de Anaconda, tiene todo de instalar, pero no tiene la interfaz visual. Se tiene que instalar todo desde comandos de terminal. Es mas practico y mucho mas liviano.

Vamos a instalar Miniconda desde la maquina virtual de Ubuntu.
En Ubuntu, colocaremos el siguiente codigo:

1.-
```Bash:
wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
```

2.- Localizas donde quedo tu archivo
```Bash:
bash Miniconda3-latest-Linux-x86_64.sh
```
En esta parte aceptar el acuerdo e instalar de forma predeterminada. Revisar donde se instalo

3.-
```Bash:
source /home/pythonuser/miniconda3/bin/activate
```
en la parte de "pythonuser" coloquen su nombre de Usuario. De esta forma activaremos Conda y aparecera un "(base)" antes de su usuario

```Bash:
(base) erikd@Errik:~$ 
```

# Instalar Miniconda en Windows

Miniconda, ya viene tambien instalado con Python por default. Si queremos usar python es la mejor forma de instalarlo, ya que viene instalado sus dependencias como pip, fundamental para instalar paquetes.

En mi caso, tengo Miniconda instalado en Windows y en mi maquina virtual de Ubuntu. 

En Ubuntu lo uso para actividades de paquetes, no necesariamente de python

En Windows son cosas meramente de Python, ya que, en windows es donde tengo instalado los editores de codigo. Podria usar el Miniconda de Ubuntu para python, pero eso implica instalar editores de codigo en mi maquina virtual

## Instalar windows

Para instalarlo es muy facil, es ir a la pagina de anaconda y descargar el miniconda para Windows

Una vez descargado hay que ejecutarlo e instalar de forma predeterminada. 

y ya tendremos nuestra Miniconda Prompt
