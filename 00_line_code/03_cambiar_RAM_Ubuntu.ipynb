# RAM de Ubuntu

### Visualizar RAM, CPU, procesos de Ubuntu

Para usar htop, hay que instalarlo con 
```Bash:
sudo apt install htop
```

```Bash:
lscpu
free -h
htop
```

La maquina virtual de Ubuntu por lo regular separa la memoria RAM a la mitad, en este caso la computadora es de 16GB de RAM y la maquina virtual, se le asignan 8GB, por esto procesos como antismash se ven limitados. Pero se puede incrementar la RAM de la maquina virtual.

### Crear archivo de texto

Se genera un archivo .txt, en donde se coloca las especificaciones, en este caso aumentar la RAM a 10GB y asignarle 6 procesadores.
```text:
[wsl2]
memory=10GB
processors=6
```
Este archivo posteriormente se debe de guardar en C:\Users\Usuario y guardarlo como .wslconfig, hay que asegurarse que se guarde como .wslconfig, sin terminacion .txt

Para comprobar, abrir archivos y en la barra superior, colocar %USERPROFILE% para dirigirse a donde debe estar el archivo. En el caso de Windows 11, en la barra superior, en medio, donde dice Ver, desplegar y poner en Mostrar y seleccionar Extensiones de nombre de archivo. Ahi se observa si es .txt o no, en caso de que si, eliminar el .txt, saldra aviso que se hara archivo inutil y aceptar.



## Reiniciar wsl2
Abrir la cmd de windows con Win+R, en la ventana que se abrio colocar cmd, una vez en la terminal cmd colocar el comando
```Windows:
wsl --shutdown
```

### Visualizar RAM y CPU
Una vez cerrado el wsl/Maquina Virtual de Ubuntu, se abrira y se correra free -h, ahi se vera que aumento la RAM. Al correr htop se vera los 6 hilos de memoria de CPU
