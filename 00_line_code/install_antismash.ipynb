# Usar AntiSmash

Trabaja con tres tipos de archivos, GenBank, Fasta y EMBL

EMBL y GenBank son genomas anotados

Descargar el archivo GenomeMining.yml: https://github.com/carpentries-incubator/genome-mining/blob/gh-pages/files/GenomeMining.yml, con las especificaciones del environment para mineria genomica.

Una vez descargado:
```Bash:
conda env create -f GenomeMining.yml

```

Tambien podemos instalar corason conda y Evomining, esos desconozco para que son, corason tiene cosas de 
BiG-SCAPE

Corason pude instalarlo sin el prefix, que viene en el codigo al crear el environment,
a lo mejor con el de genomemining tambien
