#### PSI-BLAST VS BLASTP

- BlastP simply compares a protein query to a protein database.
- PSI-BLAST allows the user to build a PSSM (position-specific scoring matrix) using the results of the first BlastP run.

(Pagina del BLAST)


#### PSI-BLAST https://www.ncbi.nlm.nih.gov/books/NBK2590/

Se usa para busqueda mas profunda en la base de datos para nuevos matches, se actualiza por interaciones subsecuentes con estas secuencias nuevas detectadas. Genera matriz de puntuacion especifica de posicion a partir de un BLASTP y con un corte de puntuaciones. 

PSI: Position-Specific Iterative. Se basa en las alineaciones generadas por un BLASTP. La primera iteracion es lo mismo que un BLASTP. 
Luego hace un MSA con los pares de mayor puntuacion del primer BLASTP por encima de cierto umbral de puntuacion o valor de E preestablecido. Luego calcula una matriz de puntuacion especifica de posicion (PSSM) a partir del MSA. PSSM captura el patron de conservacion en la alineacion. Altamente conservados con mas puntaje y poco conservadas con puntuaciones cercanas a cero. 

Luego se usa la PSSM en vez de la matriz de sustitucion.  Para detecar secuencias que matchean el patron de conservacion especificado en la PSSM. 

Las nuevas seq detectadas arriba del E-value asignado se agregan al MSA para otra ronda de busqueda. Cuando no se detectan nuevas secuencias se detiene el proceso. 

Esto hace que PSI-BLAST mas capaz de detectar secuencias distantes que un solo Query (una sola secuencia asignada en BLASTP).

Combina the underlying conservation information from a range od related sequence into a single score matrix


Valor de inclusion del Evalue de 0.005
PSI-BLAST no da una decision de si las secuencias son homologas o no.

Heuristico asumir que para proteinas globulares con un E-value <0.01 es homologo

### - criterios para usar PSIBLAST
Se uso PSI-BLAST hace una matrix de puntuacion de posicion en base a un blastP inicial, para posteriormente hacer nuevas iteraciones con esa matrix y detectar secuencias mas distantes.

### - criterios de corte del PSIBLAST
El mismo PSI-BLAST tiene un threshold de un Evalue de 0.005, ese inicialmente nos da resultados con un buen threshold del E-value. Posteriormente se selecciono un querycover del 80% y un % de similitud del 40%

### - criterios del uso de base de datos
Se uso Nrclustered, ya que junta resultados en clusters, teniendo una busqueda mas rapida y sin tener secuencias reptidas que se hayan subido

### - criterios del uso del metodo de alineamiento
la funcion de MSA solo tiene disponible los alineamientos de ClustalW, ClustalOmega y MUSCLE.

### - metodo de curation
Hay metodos en R para hacer la limpieza o un ajuste manual, modificando los valores de penalizacion de los gaps. De igual forma hay dos algoritmos muy conocidos que son el Triml y Clipkit.

## Clipkit (filtrado y eliminacion de gaps en el MSA)
## - Version 2.14.0 
Se puede hacer un reporte del trimming (https://jlsteenwyk.com/ClipKIT/advanced/index.html)
ClipKIT trims multiple sequence alignments and maintains phylogenetically informative sites.

- Cita: Steenwyk et al. 2020, PLOS Biology. doi: 10.1371/journal.pbio.3001007
https://journals.plos.org/plosbiology/article?id=10.1371/journal.pbio.3001007
- -m --mode da el modo de trimming 
smart-gap el de default
- smart-gap: dynamically determines the gap threshold by analyzing the gap distribution across the multiple sequence alignment rather than using a fixed, hard-coded percentage.
- -g --gaps: gap threshold 0.9 default
- -gc --gap_characters: specifies gap charcaters used. For aa: Xx-?*
- --ambiguity_handling: "missing" by default
- -a --auxiliary_file: Auxiliary file with metainformation for varios trimmin modes
- --report_json: write a report_json 
- --plot_trim_report: Write a html trim plot report

### modos de clipkit 
- smart-gap: dynamic determination of gaps threshold
entropy: trim sites above a normalized Shannon entropy threshold (default: 0.8)
- gappy: trim all sites that are above a threshold of gappyness (default: 0.9)
- block-gappy: trim contiguous runs of sites above a threshold of gappyness (default: 0.9)
- gappyout: infer a gap threshold from alignment-wide gap distribution and trim above it (gappyout-inspired behavior; not strict trimAl compatibility)
- composition-bias: trim sites above a normalized composition-bias threshold (default: 0.8)
- heterotachy: infer a parsimony guide tree and trim sites with high clade-to-clade entropy variation (default: 0.8)
- kpic: keep only parsimony informative and constant sites
- kpic-smart-gap: a combination of kpic- and smart-gap-based trimming
- kpic-gappy: a combination of kpic- and gappy-based trimming
- kpi: keep only parsimony informative sites
- kpi-smart-gap: a combination of kpi- and smart-gap-based trimming
- kpi-gappy: a combination of kpi- and gappy-based trimming
- c3: remove third codon position from alignment
- cst: custom site trimming (remove sites specified by the user)
