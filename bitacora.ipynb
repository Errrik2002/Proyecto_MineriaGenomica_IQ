#### PSI-BLAST https://www.ncbi.nlm.nih.gov/books/NBK2590/

Se usa para busqueda mas profunda en la base de datos para nuevos matches, se actualiza por interaciones subsecuentes con estas secuencias nuevas detectadas. Genera matriz de puntuacion especifica de posicion a partir de un BLASTP y con un corte de puntuaciones. 

PSI: Position-Specific Iterative. Se basa en las alineaciones generadas por un BLASTP. La primera iteracion es lo mismo que un BLASTP. 
Luego hace un MSA con los pares de mayor puntuacion del primer BLASTP por encima de cierto umbral de puntuacion o valor de E preestablecido. Luego calcula una matriz de puntuacion especifica de posicion (PSSM) a partir del MSA. PSSM captura el patron de conservacion en la alineacion. Altamente conservados con mas puntaje y poco conservadas con puntuaciones cercanas a cero. 

Luego se usa la PSSM en vez de la matriz de sustitucion.  Para detecar secuencias que matchean el patron de conservacion especificado en la PSSM. 

Las nuevas seq detectadas arriba del E-value asignado se agregan al MSA para otra ronda de busqueda. Cuando no se detectan nuevas secuencias se detiene el proceso. 

Esto hace que PSI-BLAST mas capaz de detectar secuencias distantes que un solo Query (una sola secuencia asignada en BLASTP).

Combina the underlying conservation information from a range od related sequence into a single score matrix


Valor de inclusion del Evalue de 0.005
PSI-BLAST no da una decision de si las secuencias son homologas o no.

Heuristico asumir que para proteinas globulares con un E-value <0.01 es homologo
